###4.1 Exploratory Data Analysis

Procedo a studiare il dataset in maniera più specifica e a pormi una serie di domande, ad esempio, mi chiedo quali siano i principali generi artistici.

Parto prendendo la colonna dei generi: in molte celle sono presenti più valori, dunque li spezzetto utilizzando come separatore il punto e virgola. Infine conto i valori, ora distribuiti su più righe.

```python
df["generi"]
```

```python
df["generi"].str.split("; ").explode().value_counts()
```

Salta subito all'occhio una prevalenza dell'arte religiosa, e una lunga serie di altri generi e sottogeneri a seguire. Per semplificare l'analisi, definisco una funzione per aggruppare i valori e successivamente la applico alla colonna.
A questo punto, aggiungo al dataframe una colonna con i generi principali.

```python
def semplifica_generi(generi):

  generi_str = str(generi).lower()

  if "arte religiosa" in generi_str:
    return "arte religiosa"
  elif any (keyword in generi_str for keyword in ["ritratto", "autoritratto", "ritratto di gruppo", "ritratto doppio"]):
    return "pittura ritrattistica"
  elif any (keyword in generi_str for keyword in ["pittura mitologica", "pittura storica", "allegoria", "portrait historié", "ritratto equestre", "dipinti di battaglie"]):
    return "pittura storica, mitologica, allegorica"
  elif "arte di genere" in generi_str:
    return "arte di genere"
  elif any(keyword in generi_str for keyword in ["pittura paesaggistica", "paesaggio urbano", "vedutismo", "paesaggio marino", "capriccio", "aeropittura"]):
    return "pittura paesaggistica"
  else:
    return "altri generi"

df["generi_semplificati"] = df["generi"].apply(semplifica_generi)

df.generi_semplificati.unique()
```

Per visualizzare i dati, genero un grafico a barre.

```python
ax = df[df["generi_semplificati"] != "altri generi"]["generi_semplificati"].value_counts().plot.bar(
    color=["#173753","#6daedb","#2892d7","#1b4353","#1d70a2"],
    figsize=(12,7),
    rot=45
    )

ax.set_facecolor('#f0f0f0')
ax.grid(color='white', linewidth=1.5, axis='y')
ax.set_axisbelow(True)
for bordo in ax.spines.values():
    bordo.set_visible(False)
colore_testo = 'black'
ax.set_title("Principali generi pittorici",
             loc="center",
             pad=20,
             fontsize=18,
             color=colore_testo)
ax.tick_params(colors=colore_testo, labelsize=11)
ax.xaxis.label.set_color(colore_testo)
ax.set_xlabel("Generi", fontsize=12)
ax.set_ylabel("Numero opere dataframe", fontsize=12, labelpad=10)
```

Voglio concentrarmi sull'arte religiosa. Creo un secondo dataframe che popolo con solo le due colonne che mi interessano, e rimuovo i valori nulli.

```python
df2 = df.filter(["data_creazione","generi_semplificati"])
df2 = df2.dropna()

df2 = df2[df2["generi_semplificati"] == "arte religiosa"].copy()
df2.isnull().sum()
```

Converto poi i valori della colonna con le date da stringhe a interi e, visto l'esteso arco temporale in esame, definisco una funzione per dividere per secoli, e la applico alla colonna in questione. Aggiungo anche qui una nuova colonna per i secoli.

```python
def raggruppa_per_secolo(anno):
  if pd.isna(anno):
    return None
  anno = int(anno)
  secolo = (anno // 100) + 1
  return f"{secolo}° secolo"
```

```python
data_creazione = df2["data_creazione"].astype(str).str.extract(r'(\d{4})', expand=False) #????????????

df2["secolo"] = pd.to_numeric(data_creazione, errors='coerce').apply(raggruppa_per_secolo)

df2.sort_values(by="secolo")
```

A questo punto procedo a indagare la tendenza che l'arte religiosa ha registrato nel corso del periodo di tempo preso in esame. Definisco quindi un'altra funzione per ordinare i secoli e ottenere una visualizzazione corretta dell'andamento dell'arte religiosa.

```python
trend_arte_religiosa_secoli = df2["secolo"].value_counts().sort_index()

ax = trend_arte_religiosa_secoli.plot(kind="line",
                                 color="#173753",
                                 figsize = (12,7),
                                 rot=0
                                )

ax.set_facecolor("#f0f0f0")
ax.grid(color="white", linewidth=1.5, axis="y")
ax.set_axisbelow(True)
for bordo in ax.spines.values():
    bordo.set_visible(False)
colore_testo = "black"
ax.set_title("Andamento dell'arte religiosa nel corso dei secoli",
             loc="center",
             pad=20,
             fontsize=18,
             color=colore_testo)
ax.tick_params(colors=colore_testo, labelsize=11)
ax.xaxis.label.set_color(colore_testo)
ax.set_xlabel("Secoli", fontsize=12)
ax.set_ylabel("Opere di arte religiosa", fontsize=12, labelpad=10)
```

###4.1 Explanatory Data Analysis

Noto un picco registrato nel XVI secolo: il contesto è quello della Riforma, e ancora di più della Controriforma. Procedo a indagare più nel dettaglio focalizzandomi su questo secolo nello specifico.

Creo un altro dataframe per il XVI secolo e visualizzo l'andamento del genere arte religiosa con un grafico.

```python
df16 = df2[df2["secolo"] == "16° secolo"].copy()
df16= df16.drop(columns="secolo")
```

```python
df16.sort_values(by="data_creazione")
```

```python
trend_16 = df16["data_creazione"].value_counts().sort_index()

ax = trend_16.plot(kind="line",
                                 color="#6daedb",
                                 figsize = (12,7),
                                 rot=0
                                )


ax.set_facecolor("#f0f0f0")
ax.grid(color="white", linewidth=1.5, axis="y")
ax.set_axisbelow(True)
for bordo in ax.spines.values():
    bordo.set_visible(False)
colore_testo = "black"
ax.set_title("Andamento dell'arte religiosa nel corso del XVI sec",
             loc="center",
             pad=20,
             fontsize=18,
             color=colore_testo)
ax.tick_params(colors=colore_testo, labelsize=11)
ax.xaxis.label.set_color(colore_testo)
ax.set_xlabel("XVI secolo", fontsize=12)
ax.set_ylabel("Opere di arte religiosa", fontsize=12, labelpad=10)
```

Il picco lo trovo a inizio secolo, una fase di transizione dal Rinascimento quattrocentesco caratterizzato da una maggiore produzione d'arte.

In seguito il grafico registra una serie di fluttuazioni e picchi ricorrenti nel corso del secolo, segnato dagli sconvolgimenti religiosi dell'Europa del tempo, tra cui:

1.   la Riforma del 1517, che ha ripercussioni sulle commissioni del genere in molte regioni del Vecchio Continente, e che si riflette sul movimento di Riforma cattolica

2.  la Controriforma e il concilio di Trento (1545-1563) con cui si sancisce un maggiore controllo da parte della Chiesa sulla produzione di arte religiosa, conscia della potenza comunicativa dall’arte, come strumento di persuasione e devozione di massa, e una codificazione del genere religioso, che porta alla pittura riformata vera e propria. Si alternano mecenatismo papale e censure ecclesiastiche più dure, come rispecchia l'andamento fluttuante del grafico.
